```python
import pandas as pd


def financial_chatbot(user_query):
    # Query ko lowercase mein convert karna
    query = user_query.lower().strip()

    # Financial Data
    data = {
        "Microsoft": {
            "total revenue": "Total Revenue for Microsoft in 2024 was $245,122 million, and in 2025 it was $278,000 million.",
            "net income": "Net Income for Microsoft in 2024 was $88,136 million, showing steady YoY growth.",
            "revenue growth": "Microsoft's revenue grew by approximately 15.67% from 2023 to 2024.",
        },
        "tesla": {
            "total revenue": "Total Revenue for Tesla in 2024 was $97,690 million.",
            "net income": "Net Income for Tesla in 2024 was $7,153 million, which experienced a YoY decline due to market conditions.",
            "revenue growth": "Tesla's revenue grew slightly by 0.95% from 2023 to 2024.",
        },
        "apple": {
            "total revenue": "Total Revenue for Apple in 2024 was $391,035 million.",
            "net income": "Net Income for Apple in 2024 was $93,736 million.",
            "revenue growth": "Apple's revenue grew by 2.02% from 2023 to 2024.",
        },
    }

    # Rule-Based Logic
    if "total revenue" in query or "revenue" in query:
        if "microsoft" in query:
            return data["Microsoft"]["total revenue"]
        elif "tesla" in query:
            return data["tesla"]["total revenue"]
        elif "apple" in query:
            return data["apple"]["total revenue"]
        else:
            return "Please specify the company name (Microsoft, Tesla, or Apple) to get total revenue details."

    elif "net income" in query or "income" in query:
        if "microsoft" in query:
            return data["Microsoft"]["net income"]
        elif "tesla" in query:
            return data["tesla"]["net income"]
        elif "apple" in query:
            return data["apple"]["net income"]
        else:
            return "Please specify the company name (Microsoft, Tesla, or Apple) to get net income details."

    elif "growth" in query:
        if "microsoft" in query:
            return data["Microsoft"]["revenue growth"]
        elif "tesla" in query:
            return data["tesla"]["revenue growth"]
        elif "apple" in query:
            return data["apple"]["revenue growth"]
        else:
            return "Please specify the company name (Microsoft, Tesla, or Apple) to get growth details."

    else:
        return "Sorry, I can only provide information on Total Revenue, Net Income, and Revenue Growth for Microsoft, Tesla, and Apple."


# Chatbot Testing / Demonstration
test_queries = [
    "What is the total revenue for Microsoft?",
    "How has Net Income changed for Tesla?",
    "Show me Apple revenue growth.",
    "What is the stock price of Google?",
]

print("--- Financial Chatbot Test Demonstration ---\n")
for q in test_queries:
    print(f"User Query: {q}")
    print(f"Chatbot Response: {financial_chatbot(q)}")
    print("-" * 60)
```

# Financial Chatbot Prototype Documentation

## Logic & Implementation
- **Rule-based Logic:** The chatbot uses conditional statements (`if-elif-else`) to parse predefined financial queries regarding Total Revenue, Net Income, and Revenue Growth for Microsoft, Tesla, and Apple.
- **Error Handling:** If a user asks a query outside the predefined financial parameters, the chatbot gracefully responds with a clear guide on supported topics.
